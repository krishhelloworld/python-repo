```python
import matplotlib.pyplot as plt 
import numpy as np 
import sys
```

```python
x = np.linspace(1,10,100)
y = np.exp(x)
```

```python
x = np.linspace(1e-45, 1, 1000)
y = np.log(x)

# Calculate the exact x where y = -100,000 (x = e^-100000)
target_y = -100000
target_x = np.exp(target_y)
print(f"To reach y = {target_y}, x must be approximately: {target_x}")

plt.figure(figsize=(9, 5))
plt.plot(x, y, color='blue', linewidth=2, label=r'$y = \ln(x)$')

# Force the y-axis to show down to -100,000
plt.ylim(-100000, 5)
plt.xlim(0, 1)

plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.title(r'Log Graph with Y-Axis Extended Down to $-100,000$')
plt.xlabel('x')
plt.ylabel('y = ln(x)')
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()

plt.show()
```
